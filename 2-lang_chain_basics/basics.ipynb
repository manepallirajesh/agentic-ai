```python
from langchain_groq import ChatGroq

llm_qroq = ChatGroq(model='qwen-qwq-32b')
llm_qroq.invoke('Hi, How are you?')
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate([
    ('system', 'You are considered as Chartered Accountant expertise, provide  the answer'),
    ('user', '{input}')
])


prompt
```

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt|llm_qroq|StrOutputParser()

response = chain.invoke({'input': 'does India gets benefited with GST?'})

print(response)
```

```python
from langchain_core.output_parsers import JsonOutputParser
json_mode_groq = ChatPromptTemplate(
    [
        ('system', 'You as an AI Engineer Expert, privude output in Json format'),
        ('user', '{input}')
    ]
)

json_chain = json_mode_groq|llm_qroq|JsonOutputParser()
print(json_chain.invoke({'input': 'Provide alternate versions of similar langhcain tools for on-prem'}))
```

```python
from pydantic import BaseModel, Field

class ProductQuery(BaseModel):
    name: str = Field(description= 'define product name')
    details: str = Field(description='provide more details')
    price: float = Field(description='provides tentative price')

json_model_groq = ChatPromptTemplate(
    [
        ('system', 'You provide information for the give productname in json output format - name:, details:, price:'),
        ('user', '{input}')
    ]
)
output_parser = JsonOutputParser(pydantic_object=ProductQuery)
chain = json_mode_groq|llm_qroq|output_parser
chain.invoke({'input':'iphone 16'})
```
